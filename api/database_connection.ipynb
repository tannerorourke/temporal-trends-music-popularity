```python
import sqlite3

db = sqlite3.connect("spotify.db")

# ------ CREATE TABLE -----------------
try:
    cursor = db.cursor()
    
    cursor.execute('''CREATE TABLE analysis(
                    id INTEGER PRIMARY KEY,
                    uri VARCHAR(100),
                    rel_date VARCHAR(100),
                    explicit INT,
                    time_signature INT,
                    key INT, 
                    tempo DECIMAL, 
                    mode INT, 
                    duration INT, 
                    danceability DECIMAL, 
                    valence DECIMAL,
                    energy DECIMAL, 
                    loudness DECIMAL,
                    speechiness DECIMAL,
                    acousticness DECIMAL,
                    instrumentalness INT, 
                    liveness DECIMAL)''')
    
    db.commit()
    db.close()
except Exception as e:
    print(e)

#---------- db.execute('''DROP TABLE audio_features''') -----------
```
